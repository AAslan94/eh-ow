```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams.update({
    "text.usetex": True,
    "font.family": "serif",
    "font.serif": ["Palatino"],
    "font.size": 18,
    "lines.linewidth": 2,
})

results_nom = np.load("area_only_cma_results_A2.npy", allow_pickle=True)
results_rob = np.load("robust_minimax_results_A2.npy")  # [X,Y,Z,...,Area(m2),...]


# ---- Nominal data ----
x_nom = np.array([res['pos'][0] for res in results_nom])
areas_nom = np.array([res['area_cm2'] for res in results_nom])

# ---- Robust data ----
x_rob = results_rob[:, 0]
theta_rob = np.degrees(results_rob[:, 3])
phi_rob = np.degrees(results_rob[:, 4])
areas_rob = results_rob[:, 5] * 1e4  # m² → cm²

# ---- Sort for clean curves ----
idx_nom = np.argsort(x_nom)
idx_rob = np.argsort(x_rob)

x_nom = x_nom[idx_nom]
areas_nom = areas_nom[idx_nom]

x_rob = x_rob[idx_rob]
areas_rob = areas_rob[idx_rob]
theta_rob = theta_rob[idx_rob]
phi_rob = phi_rob[idx_rob]

# =========================
# ---- Plot 1: Area ----
# =========================
fig1, ax_area = plt.subplots(figsize=(8, 5))

ax_area.plot(x_nom, areas_nom,
             marker='o', linestyle='-', color='blue', label='Default')
ax_area.plot(x_rob, areas_rob,
             marker='s', linestyle='--', color='red', label='Robust')

ax_area.set_ylabel(r'$A_\mathrm{PV}$ (cm$^2$)')
ax_area.set_xlabel(r'$x$ [m]')
ax_area.grid(True, linestyle='--', alpha=0.7)
ax_area.legend(loc='best', frameon=True)

plt.tight_layout()
plt.savefig("area_opt_A2.pdf", format="pdf", bbox_inches="tight")
plt.show()
plt.close(fig1)


# =========================
# ---- Plot 2: Theta & Phi ----
# =========================
fig2, ax_theta = plt.subplots(figsize=(8, 5))

# Theta (left y-axis)
line_theta = ax_theta.plot(x_rob, theta_rob,
                           marker='^', linestyle='-', color='green',
                           label=r'$\theta$')
ax_theta.set_ylabel(r'$\theta$ (deg)', color='green')
ax_theta.set_ylim(40, 80)
ax_theta.tick_params(axis='y', labelcolor='green')
ax_theta.set_xlabel(r'$x$ [m]')
ax_theta.grid(True, linestyle='--', alpha=0.7)

# Phi (right y-axis)
ax_phi = ax_theta.twinx()
line_phi = ax_phi.plot(x_rob, phi_rob,
                       marker='d', linestyle='-', color='purple',
                       label=r'$\phi$')
ax_phi.set_ylabel(r'$\phi$ (deg)', color='purple')
ax_phi.tick_params(axis='y', labelcolor='purple')

lines = line_theta + line_phi
labels = [l.get_label() for l in lines]
ax_theta.legend(lines, labels, loc='lower right', frameon=True)

plt.tight_layout()
plt.savefig("theta_phi_opt_A2.pdf", format="pdf", bbox_inches="tight")
plt.show()
plt.close(fig2)
```

```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams.update({
    "text.usetex": True,
    "font.family": "serif",
    "font.serif": ["Palatino"],
    "font.size": 18,
    "lines.linewidth": 2,
})

# ---- Load ----
results_nom = np.load("area_only_cma_results_A2.npy", allow_pickle=True)
results_rob = np.load("robust_minimax_results_A2.npy")
results_cma = np.load("cma_simple_A2.npy", allow_pickle=True)

# ---- Nominal ----
x_nom = np.array([r['pos'][0] for r in results_nom])
area_nom = np.array([r['area_cm2'] for r in results_nom])

# ---- CMA ----
x_cma = np.array([r['pos'][0] for r in results_cma])
area_cma = np.array([r['area_cm2'] for r in results_cma])

# ---- Robust ----
x_rob = results_rob[:,0]
theta_rob = np.degrees(results_rob[:,3])
phi_rob = np.degrees(results_rob[:,4])
area_rob = results_rob[:,5] * 1e4   # m² → cm²

# ---- Sort along diagonal ----
s_nom = np.argsort(x_nom)
s_cma = np.argsort(x_cma)
s_rob = np.argsort(x_rob)

x_nom, area_nom = x_nom[s_nom], area_nom[s_nom]
x_cma, area_cma = x_cma[s_cma], area_cma[s_cma]

x_rob = x_rob[s_rob]
area_rob = area_rob[s_rob]
theta_rob = theta_rob[s_rob]
phi_rob = phi_rob[s_rob]

# ---- Fix φ wrap ----
phi_rob = np.where(phi_rob > 359.5, phi_rob - 360, phi_rob)

# ---- Plot ----
fig, ax = plt.subplots(3,1, figsize=(8,12), sharex=True)

# Area
ax[0].plot(x_nom, area_nom, 'o-', label="Nominal")
ax[0].plot(x_cma, area_cma, '^-', label="CMA")
ax[0].plot(x_rob, area_rob, 's--', label="Robust")

ax[0].set_ylabel(r'Area (cm$^2$)')
ax[0].grid(True, linestyle='--', alpha=0.7)
ax[0].legend()

# Theta
ax[1].plot(x_rob, theta_rob, 'd-', color='green')
ax[1].set_ylabel(r'$\theta$ (deg)')
ax[1].grid(True, linestyle='--', alpha=0.7)

# Phi
ax[2].plot(x_rob, phi_rob, 'o-', color='purple')
ax[2].set_ylabel(r'$\phi$ (deg)')
ax[2].set_xlabel(r'Diagonal coordinate [m]')
ax[2].grid(True, linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```
